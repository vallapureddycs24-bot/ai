```python
import heapq

# Function to calculate the number of misplaced tiles (heuristic h)
def misplaced_tiles(state, goal):
    return sum(1 for i in range(len(state)) if state[i] != 0 and state[i] != goal[i])

# Function to get possible moves from the current state
def get_neighbors(state):
    neighbors = []
    size = 3  # 3x3 puzzle
    zero_index = state.index(0)
    row, col = divmod(zero_index, size)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # up, down, left, right
    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < size and 0 <= new_col < size:
            new_index = new_row * size + new_col
            new_state = list(state)
            new_state[zero_index], new_state[new_index] = new_state[new_index], new_state[zero_index]
            neighbors.append(tuple(new_state))
    return neighbors

# A* Search Algorithm
def a_star(start, goal):
    open_list = []
    heapq.heappush(open_list, (0, start, 0))  # (f, state, g)
    came_from = {}
    g_score = {start: 0}

    while open_list:
        f, current, g = heapq.heappop(open_list)

        # Display current state with g, h, f
        h = misplaced_tiles(current, goal)
        print("State:")
        for i in range(0, 9, 3):
            print(current[i:i+3])
        print(f"g = {g}, h = {h}, f = {f}\n")

        if current == goal:
            print("Goal reached!")
            return

        for neighbor in get_neighbors(current):
            tentative_g = g + 1
            if neighbor not in g_score or tentative_g < g_score[neighbor]:
                g_score[neighbor] = tentative_g
                h_val = misplaced_tiles(neighbor, goal)
                f_val = tentative_g + h_val
                heapq.heappush(open_list, (f_val, neighbor, tentative_g))
                came_from[neighbor] = current

# Main execution
if __name__ == "__main__":
    try:
        # User input for start and goal states
        print("Enter start state (9 numbers, use 0 for blank):")
        start_state = tuple(map(int, input().split()))
        print("Enter goal state (9 numbers, use 0 for blank):")
        goal_state = tuple(map(int, input().split()))

        if len(start_state) != 9 or len(goal_state) != 9:
            raise ValueError("Each state must have exactly 9 numbers.")

        a_star(start_state, goal_state)

    except ValueError as e:
        print("Error:", e)
```
